## Exercise 1: Shape polymorphism
Build a `Shape` base class and `Circle`/`Rectangle` subclasses. Give each an `area()` method (override it). Store shapes in a list and print every area in a loop (polymorphism).

```python
import math

class Shape:
    """Base class for all shapes."""

    def area(self):
        raise NotImplementedError("Subclasses must implement area()")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height
```

```python
shapes = [Circle(3), Rectangle(4, 5), Circle(1.5), Rectangle(2, 3)]

for shape in shapes:
    # the loop doesn't care whether it's a Circle or Rectangle -
    # it just calls .area() and trusts each class to do it correctly
    print(f"{type(shape).__name__} area: {shape.area():.2f}")
```

## Exercise 2: BankAccount encapsulation
Create a `BankAccount` with a private `__balance`. Provide `deposit(amount)` and `withdraw(amount)` methods that validate (no negative deposits, no overdraft). Expose `get_balance()`.

```python
class BankAccount:
    """A bank account with a private balance."""

    def __init__(self, starting_balance=0):
        # double underscore -> name-mangled to _BankAccount__balance
        self.__balance = starting_balance

    def deposit(self, amount):
        if amount <= 0:
            print(f"Deposit failed: amount must be positive (got {amount})")
            return
        self.__balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            print(f"Withdrawal failed: amount must be positive (got {amount})")
            return
        if amount > self.__balance:
            print(f"Withdrawal failed: insufficient funds (balance is {self.__balance})")
            return
        self.__balance -= amount

    def get_balance(self):
        return self.__balance
```

```python
account = BankAccount(1000)

account.deposit(500)
account.withdraw(200)
account.withdraw(5000)   # should fail, insufficient funds
account.deposit(-50)     # should fail, negative deposit

print(f"Final balance: {account.get_balance()}")
```

## Exercise 3: Book alternative constructor
Write a `@classmethod` alternative constructor `Book.from_string("title|author")` that splits the string and returns a `Book`.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    @classmethod
    def from_string(cls, raw_string):
        # 'cls' is the Book class itself here, not an instance,
        # so this method can build and return a new Book
        title, author = raw_string.split("|")
        return cls(title.strip(), author.strip())

    def describe(self):
        print(f"{self.title} by {self.author}")
```

```python
normal_book = Book("Book1", "Authur1")
normal_book.describe()

raw_book_strings = ["Book2|Authur2", "Book3|Authur3"]
for raw in raw_book_strings:
    book = Book.from_string(raw)
    book.describe()
```

### What I Learned
Polymorphism means the calling code (the `for shape in shapes` loop) never needs to know which exact subclass it's dealing with. Encapsulation with `__balance` is stronger than the single-underscore convention, since it's actually name-mangled by Python. `@classmethod` gives an alternative entry point into building an object, useful when the input data doesn't match the normal constructor's shape.
