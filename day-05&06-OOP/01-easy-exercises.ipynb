## Exercise 1: Book class
Create a `Book` class with instance attributes `title` and `author`, and a method `describe()` that prints `"<title> by <author>"`. Instantiate two books and call `describe()` on each.

```python
class Book:
    """A book with a title and author."""

    def __init__(self, title, author):
        # instance attributes - unique to each book
        self.title = title
        self.author = author

    def describe(self):
        print(f"{self.title} by {self.author}")
```

```python
book1 = Book("Book1", "Author1")
book2 = Book("Book2", "Author2")

book1.describe()
book2.describe()
```

## Exercise 2: Class attribute counter
Add a class attribute `count` to `Book` that increments in `__init__` every time a book is created. Print the total after making three books.

```python
class Book:
    """A book with a title and author, tracking how many have been made."""

    # class attribute - shared by every Book, tracks total books created
    count = 0

    def __init__(self, title, author):
        self.title = title
        self.author = author
        Book.count += 1

    def describe(self):
        print(f"{self.title} by {self.author}")
```

```python
b1 = Book("Book1", "Author1")
b2 = Book("Book2", "Author2")
b3 = Book("Book3", "Author3")

print(f"Total books created: {Book.count}")
print("The books are:")
b1.describe()
b2.describe()
b3.describe()
```

### What I Learned
A class attribute like `count` lives on the class itself, not on any one object, so every `Book` instance updating it via `Book.count += 1` is really updating one shared value. This is different from `self.title`, which is a fresh instance attribute for every object.
