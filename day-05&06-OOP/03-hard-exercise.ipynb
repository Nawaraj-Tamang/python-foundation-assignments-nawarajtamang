## Exercise 1: Multilevel inheritance
Implement multilevel inheritance `Vehicle -> Car -> SportsCar`, each constructor using `super().__init__(...)`. The final object should carry attributes from all three levels.

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand


class Car(Vehicle):
    def __init__(self, brand, color):
        super().__init__(brand)  # sets self.brand
        self.color = color


class SportsCar(Car):
    def __init__(self, brand, color, top_speed):
        super().__init__(brand, color)  # sets self.brand and self.color
        self.top_speed = top_speed

    def describe(self):
        return (f"{self.brand} sports car, {self.color} color, "
                f"top speed {self.top_speed} km/h")
```

```python
car = SportsCar("Ferrari", "Red", 340)

# every attribute is present, even though each was set at a
# different level of the inheritance chain
print(car.brand)
print(car.color)
print(car.top_speed)
print(car.describe())
```

## Exercise 2: Payment abstraction
Define an abstract `PaymentMethod(ABC)` with `@abstractmethod def pay(self, amount)`. Implement `CreditCard`, `Esewa`, and `MobileBanking`. Process a list of payments polymorphically. Verify that instantiating `PaymentMethod()` directly raises `TypeError`.

```python
from abc import ABC, abstractmethod

class PaymentMethod(ABC):
    """Abstract base class - defines the contract every payment method must follow."""

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(PaymentMethod):
    def pay(self, amount):
        print(f"Charging {amount} to credit card")


class Esewa(PaymentMethod):
    def pay(self, amount):
        print(f"Sending {amount} via eSewa")


class MobileBanking(PaymentMethod):
    def pay(self, amount):
        print(f"Transferring {amount} via mobile banking")
```

```python
payments_to_process = [
    (CreditCard(), 2500),
    (Esewa(), 850),
    (MobileBanking(), 1200),
]

for handler, amount in payments_to_process:
    # same call, .pay(amount), dispatches differently per class -
    # that's polymorphism working together with abstraction
    handler.pay(amount)
```

```python
# confirm the abstract class truly cannot be instantiated directly
try:
    PaymentMethod()
except TypeError as error:
    print(f"As expected, this failed: {error}")
```

### What I Learned
Chaining `super().__init__()` across three levels means each class only needs to know about the level directly above it, not the whole chain. `ABC` plus `@abstractmethod` is a much stronger guarantee than convention: Python refuses to even create an instance of `PaymentMethod`, or any subclass that forgets to implement `pay()`.
